```python
# 1. FILOSOFÍA DE PYTHON (ZEN DE PYTHON)
import this

# 2. SIMULACIÓN DE OPERACIONES EN BASE DE DATOS
import sqlite3

# Conexión a la base de datos local
conexion = sqlite3.connect("mi_base_de_datos.db")
cursor = conexion.cursor()

# Crear tabla de prueba
cursor.execute("""
CREATE TABLE IF NOT EXISTS usuarios (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre TEXT NOT NULL,
    email TEXT NOT NULL
)
""")

print("--- OPERACIONES EN LA BASE DE DATOS ---")

# A. INSERTAR (Crear registro)
cursor.execute("INSERT INTO usuarios (nombre, email) VALUES ('Carlos Pérez', 'carlos@example.com')")
conexion.commit()
print("✓ Registro insertado con éxito.")

# B. CONSULTAR (Leer registros)
cursor.execute("SELECT * FROM usuarios")
registros = cursor.fetchall()
print("\nRegistros en la base de datos:")
for fila in registros:
    print(fila)

# C. ACTUALIZAR (Modificar registro)
cursor.execute("UPDATE usuarios SET email = 'carlos_nuevo@example.com' WHERE nombre = 'Carlos Pérez'")
conexion.commit()
print("\n✓ Registro actualizado con éxito.")

# D. ELIMINAR (Borrar registro)
cursor.execute("DELETE FROM usuarios WHERE nombre = 'Carlos Pérez'")
conexion.commit()
print("✓ Registro eliminado con éxito.")

# Cierre de conexión
conexion.close()
print("\n✓ Proceso finalizado correctamente.")
```
